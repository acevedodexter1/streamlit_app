# GenAI MVP Lab: Planning as a Data Scientist
In this lab, you'll use Generative AI as a co-creator to plan a Minimum Viable Product (MVP) for a data-driven app. Instead of working alone, you'll collaborate with GenAI to brainstorm features, prioritize tasks, and map out a build plan.
You will do just like Priya, an ML engineer at a healthcare startup, who used GenAI to scope and refine her MVP for extracting insights from clinical trial PDFs.

## Step 1: Define Your Prompt
Use this template to frame your request to GenAI:

You are a product manager with experience in healthcare data apps.
I need help with planning an MVP for an AI-powered medicine reminder and interaction checker.
My situation is I am a data science student who needs to build a working prototype that takes a list of patient medications and outputs a daily schedule and potential drug interaction warnings.
Please help me outline a build plan in bullet points.

- End Goal: Build a simple Streamlit web layout that accepts a text description of a user's movie mood and outputs 3 tailored film recommendations.
- Target Audience: Casual movie watchers looking for instant, personalized inspiration without endless scrolling.
- Metrics for Success: Fast execution (under 3 seconds) with clean UI formatting and no generation crashes.

### Step 4: Domain and Audience Analysis
Domain: Healthcare & Patient Safety (Medication Management)

Data Problem: Patients don't know if their medications are safe to take together or forget their schedules.

Target User: Elderly patients, caregivers, or people with maintenance medications.

Success Criteria: Generate a daily schedule and provide drug interaction warnings in under 3 seconds.

Current Frustrations: Confusing prescriptions, hard to remember times, and no easy way to check interactions.

Deployment Platform: A simple Streamlit web app.


### Step 5: MVP Success Metrics 
Behavioral Metric: Percentage of users returning daily to check their schedule.

Business Metric: User retention rate (percentage of repeat users visiting the app).

Qualitative Metric: "Thumbs Up/Down" feedback on whether the interaction warning was helpful.

Leading Indicator: Number of medicines entered into the system in the first week.


### Step 6: Prioritize Features 
Must-Have (Core MVP): Input form for medicine, daily schedule display, and an AI chatbot for basic medicine questions.

Nice-to-Have (Next Version): SMS/Email notifications, refill reminders, and a printable schedule.

## Final Reminder
Generative AI is great for:
- Brainstorming features and workflows
- Generating boilerplate code and queries
- Debugging and planning
- Recommending architecture and tools

But you’ll need human judgment for:
- Picking the right problem to solve
- Making UX decisions for your users
- Applying business logic
- Validating and interpreting AI outputs
